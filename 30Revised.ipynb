```python
operation = int(input("Enter 1 for sum, 2 for sub, 3 for mul, 4 for div, 5 for modular, 6 for square"))
a = float(input("Enter the first number"))
b = float(input("Enter the second number"))

if operation==1:
    print(f"Sum = {a+b}")

elif operation==2:
    print(f"Sub = {a-b}")
elif operation==3:
    print(f"Mul = {a*b}")
elif operation==4:
    print(f"Div = {a/b}")
elif operation==5:
    print(f"Moduler = {a%b}")
elif operation==6:
    print(f"Square = {a**b}")
else:
    print("Input the valid number")

```

```python
#print the all odd numbers between 1 to 20

for i in range(21):
    if i%2 !=0:
        print(i)

#print the table of 57

for i in range(11):
    print(i*57)

#print all the multiples 3 between 1-50 but skip 15 and 21


for i in range(1,50):
    if i == 15 or i==21:
        continue
    elif i%3==0:
        print(i)
```

List, Tuple, Dict, set

```python
marks = [23,56,34,212,43,'a','b',5.66]
print(marks, type(marks))
# print(marks.find(a))
print(marks[0:4])#print 0-4 index from the list
print(marks[5])#print just 5th index value
print(marks[-1])#negative index for last value
print(marks[-5:-1])#negative slicing
print(marks[-5:])
print(marks[-1:-4:-1])#backward slicing. in backward slicing have add the step
```

```python
revenue = [10,15,40,70.80]
expense = [5,7,30,90,70]

for index in range(len(revenue)):
    print(index, end=" ")
for index in range(len(expense)):
    print(index, end=" ")
```
